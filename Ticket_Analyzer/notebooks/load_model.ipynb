## **Customer Support Ticket Classification,NLP and Multi-Model Pipeline**

**Description:** Loads trained NLP models and vectorizers to categorize and analyze new incoming support tickets.

**Import Libraries**

```python
import joblib
import pandas as pd
import sys

from pathlib import Path
BASE_DIR = Path.cwd().parent if Path.cwd().name == "notebooks" else Path.cwd()
src_path = str(BASE_DIR / "src")

if src_path not in sys.path:
    sys.path.append(src_path)

from utils import LoadData, Eda, TextPreprocessor, FeatureEngineer, TrainModel, SavingModel

print("Done!")
```

**Path Setup**

```python
model_path = str(BASE_DIR / "model" / "best_ticket_classifier.pkl")
model = joblib.load(model_path)
```

**Data Loading**

```python
data = pd.read_csv(str(BASE_DIR / "data" / "customer_support_tickets.csv"))
```

**Text Preprocessing**

```python
preprocessing = TextPreprocessor()
data['cleaned_text'] = data['Ticket Description'].apply(preprocessing.clean_text)
```

**Load Model and Saved Vectorizer**

```python
trained_model = model['model']
vectorizer = model['vectorizer']
```

**Vectorize Cleaned Text**

```python
X = vectorizer.transform(data['cleaned_text'])
```

**Run Predictions**

```python
predictions = trained_model.predict(X)
```

```python
data['Predicted_Result'] = predictions
print(data[['Ticket ID', 'Ticket Subject', 'Predicted_Result']].head())
```
